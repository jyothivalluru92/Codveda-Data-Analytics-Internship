### Dataset: Social Media Sentiment Analysis
**Objective:** Clean the raw data by handling missing values, removing duplicates, and standardizing text for analysis.

```python
# Importing pandas for data manipulation
!pip install pandas
import pandas as pd

# Load the data - using relative path to reach the Datasets folder
df = pd.read_csv('../../Dataset/Sentiment dataset.csv')

# Display the first 5 rows
df.head()
```

## 1. Initial Data Inspection
We check for the shape of the data, column types, and any immediate issues like "Unnamed" columns.

```python
print(f"Dataset Shape: {df.shape}")
print("\nColumn Info:")
df.info()
print("\nMissing Values:")
print(df.isnull().sum())
```

## 2. Data Cleaning
1. Remove unnecessary index columns.
2. Clean leading/trailing whitespaces from text and category columns.
3. Handle potential missing values.

```python
# 1. Drop unnecessary columns
df_cleaned = df.drop(columns=['Unnamed: 0.1', 'Unnamed: 0'])

# 2. Standardize string columns (Remove spaces and convert to lowercase)
df_cleaned['Sentiment'] = df_cleaned['Sentiment'].str.strip().str.capitalize()
df_cleaned['Platform'] = df_cleaned['Platform'].str.strip()

# 3. Convert Timestamp to datetime objects
df_cleaned['Timestamp'] = pd.to_datetime(df_cleaned['Timestamp'])

# 4. Remove duplicate rows
original_count = len(df_cleaned)
df_cleaned = df_cleaned.drop_duplicates()
final_count = len(df_cleaned)

print(f"Removed {original_count - final_count} duplicate rows.")
print(f"New dataset shape: {df_cleaned.shape}")

# Show cleaned info
df_cleaned.head()
```

## 3. Data Filtering & Extraction
Extracting specific insights, such as filtering for high-engagement posts on a specific platform.

```python
# Example: Filter for 'Positive' sentiments on 'Instagram' with more than 20 likes
high_engagement_insta = df_cleaned[(df_cleaned['Platform'] == 'Instagram') & 
                                   (df_cleaned['Sentiment'] == 'Positive') & 
                                   (df_cleaned['Likes'] > 20)]

print(f"Found {len(high_engagement_insta)} high-engagement Instagram posts.")
high_engagement_insta.head()
```

## 4. Final Verification
We now verify that there are no remaining missing values, no duplicates, and that the data types are correct.

```python
# 1. Check for missing values one last time
print("Missing values after cleaning:")
print(df_cleaned.isnull().sum())

# 2. Check for duplicates
print(f"\nDuplicate rows remaining: {df_cleaned.duplicated().sum()}")

# 3. Confirm data types (especially for the 'Timestamp')
print("\nFinal Data Types:")
print(df_cleaned.dtypes)
```

## 5. Exporting the Cleaned Dataset
Saving the cleaned data to a new CSV file for use in future tasks.

```python
# Export to the current Task 1 folder
df_cleaned.to_csv('cleaned_sentiment_data.csv', index=False)
print("File 'cleaned_sentiment_data.csv' has been saved successfully!")
```
