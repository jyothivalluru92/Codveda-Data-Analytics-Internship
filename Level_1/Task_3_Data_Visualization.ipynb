### Dataset: Iris Flower Dataset
**Objective:** Create professional Bar Plots, Line Charts, and Scatter Plots. Customize all visual elements and export the results as images for reporting.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load the Iris dataset
df = pd.read_csv('../../Dataset/iris.csv')

# Set aesthetic style
sns.set_theme(style="whitegrid")

# Create a folder for exported images if it doesn't exist
if not os.path.exists('plots'):
    os.makedirs('plots')

df.head()
```

## 1. Bar Plot: Average Measurements per Species
This chart compares the mean petal length across the three Iris species. We use custom labels and titles as per requirements.

```python
plt.figure(figsize=(10, 6))
# Create bar plot
sns.barplot(
    data=df,
    x='species',
    y='petal_length',
    hue='species',
    palette='viridis',
    errorbar=None,
    legend=False
)

# --- CUSTOMIZATION ---
plt.title('Average Petal Length by Iris Species', fontsize=14, fontweight='bold')
plt.xlabel('Flower Species', fontsize=12)
plt.ylabel('Average Petal Length (cm)', fontsize=12)

# --- EXPORTING ---
plt.savefig('plots/1_bar_plot.png', dpi=300)
plt.show()
```

## 2. Line Chart: Sample Variation
A line chart showing the variation of Sepal Length and Petal Length across all samples in the dataset.

```python
plt.figure(figsize=(12, 6))

# Plotting lines
plt.plot(df['sepal_length'], label='Sepal Length', color='teal', linewidth=1.5)
plt.plot(df['petal_length'], label='Petal Length', color='orange', linewidth=1.5)

# --- CUSTOMIZATION ---
plt.title('Trend of Sepal vs Petal Length across Samples', fontsize=14)
plt.xlabel('Sample Index', fontsize=12)
plt.ylabel('Measurement (cm)', fontsize=12)
plt.legend(title='Features', loc='upper right') # Custom Legend

# --- EXPORTING ---
plt.savefig('plots/2_line_chart.png', dpi=300)
plt.show()
```

## 3. Scatter Plot: Identifying Clusters
Visualizing the relationship between Sepal Length and Sepal Width. This is essential for understanding how species cluster together.

```python
plt.figure(figsize=(10, 6))

# Create scatter plot with hue for species
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species', style='species', s=100)

# --- CUSTOMIZATION ---
plt.title('Scatter Plot: Sepal Length vs Sepal Width', fontsize=14)
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') # Legend placement

# --- EXPORTING ---
plt.savefig('plots/3_scatter_plot.png', dpi=300)
plt.show()
```
